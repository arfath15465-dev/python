### UNDERSTOOD 2D AND 1D ARRAY AND PERFORMS THE OPERATION 

```python
print("This is a vector i.e 1D Array")
vector = np.array([1,2,3])
print(vector)
print("This is a matrix i.e 2D Array")
matrix = np.array([[1,2,3],[4,5,6]])
print(matrix)
```

```python
print(vector * matrix)
```

### Create a (3,2) shaped 2d array

```python
print("This is a matrix i.e (3,2) 2D Array")
matrix1 = np.array([[1,2]])
matrix2 = np.array([[1,2],[3,4],[5,6]])
print(matrix1*matrix2)
```
