### Apply Function

### Define a function that return
* If marks greater than 80 return A
* If marks greater than 60 return B
* If marks greater than 40 return C
* return fail

```python
def grade_function(marks):
    if marks >80:
        return 'A'
    elif marks >60:
            return 'B'
    elif marks >40:
            return 'C'
    else:
          return 'Fail'
print(grade_function(100))
    
```

```python
df['Grade']=df['MARKS'].apply(grade_function)
df
```

```python
df['Bonus']=df['MARKS'].apply(lambda x:x+5)
df
```

### JOIN IN PYTHON (MERGE)

```python
students = pd.DataFrame({'student_id':[1,2,3,6],'name':['Shahzeb','Ammar','Abdullah','Adnan']})
marks = pd.DataFrame({'student_id':[1,2,3,4],'marks':[100,50,20,50]})
```

```python
pd.merge(
    students,
    marks,
    on='student_id',
    how='inner'
)
```

```python
pd.merge(
    students,
    marks,
    on='student_id',
    how='right'
)
```

```python
pd.merge(
    students,
    marks,
    on='student_id',
    how='left'
)
```

```python
s = pd.DataFrame({'student_id':[1,2,3,6],'name':['Shahzeb','Ammar','Abdullah','Adnan']})
m = pd.DataFrame({'id':[1,2,3,4],'marks':[100,50,20,50]})
```

```python
d=pd.merge(
    s,
    m,
    left_on='student_id',
    right_on='id',
    how='left'
)
d[['student_id','marks']]
```
